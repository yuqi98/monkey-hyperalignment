# Movie repetition analysis (Supp. Fig. S7, monkey only)

Loads the ISC-by-repetition grids computed by `compute_isc_diff_rep.py` / `assemble_isc_grid_diff_rep.py`.

```python
import numpy as np
import matplotlib.pyplot as plt

ISC_ROOT = "/dartfs/rc/lab/H/HaxbyLab/yuqi/monkey_kingdom_data/rep_analysis/isc"

MAX_REPS = 7
reps = np.arange(1, MAX_REPS + 1)

anatomical = np.load(f"{ISC_ROOT}/anatomical_avg_isc_by_rep.npy")  # (7,)
ridge_grid = np.load(f"{ISC_ROOT}/ridge_avg_isc_grid.npy")  # (7, 7): [transform_rep-1, data_rep-1]
```

## Supp. Fig. S7b: ISC by (transform repetitions, data repetitions), ridge

```python
ridge_diag = np.diag(ridge_grid)

fig, ax = plt.subplots(figsize=(6.5, 5.5))
im = ax.imshow(ridge_grid, origin="lower", cmap="viridis", aspect="auto")
ax.set_xticks(range(MAX_REPS))
ax.set_xticklabels(reps)
ax.set_yticks(range(MAX_REPS))
ax.set_yticklabels(reps)
ax.set_xlabel("Repetitions used for test data", fontsize=12)
ax.set_ylabel("Repetitions used to build transform", fontsize=12)
ax.set_title("Ridge", fontsize=15)
for i in range(MAX_REPS):
    for j in range(MAX_REPS):
        ax.text(j, i, f"{ridge_grid[i, j]:.2f}", ha="center", va="center",
                 color="white" if ridge_grid[i, j] < ridge_grid.mean() else "black", fontsize=9)
fig.suptitle("Average ISC across repetition counts", fontsize=16)
fig.colorbar(im, ax=ax, shrink=0.85, label="Average ISC (r)")
plt.savefig("isc_diff_rep_heatmap_ridge_only.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Supp. Fig. S7a: anatomical vs. ridge, matched repetition count

```python
plt.figure(figsize=(7, 5.5))
plt.plot(reps, anatomical, "-o", color="#888888", label="Anatomical", lw=2, markersize=7)
plt.plot(reps, ridge_diag, "-o", color="#E31A1C", label="Ridge (matched rep)", lw=2, markersize=7)
plt.xlabel("Repetitions used per clip", fontsize=12)
plt.ylabel("Average ISC (r)", fontsize=12)
plt.title("ISC by repetition count: anatomical vs ridge", fontsize=14)
plt.xticks(reps)
plt.legend(fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.savefig("isc_diff_rep_matched_comparison_ridge_only.png", dpi=300, bbox_inches="tight")
plt.show()
```
